```python
import scipy.io
import pydicom
import numpy as np
import matplotlib.pyplot as plt
import os

output_img = "/content/pe_dataset/images"
output_mask = "/content/pe_dataset/masks"

os.makedirs(output_img, exist_ok=True)
os.makedirs(output_mask, exist_ok=True)

patients = os.listdir(ct_path)

for patient in patients:
    print("Processing:", patient)

    dcm_folder = os.path.join(ct_path, patient)
    mat_file = os.path.join(mask_path, patient + ".mat")

    # Load mask
    mat = scipy.io.loadmat(mat_file)
    mask = mat['Mask']

    dcm_files = sorted(os.listdir(dcm_folder))

    for i, file in enumerate(dcm_files):
        if file.endswith(".dcm"):

            ds = pydicom.dcmread(os.path.join(dcm_folder, file))
            image = ds.pixel_array

            # Normalize
            image = (image - np.min(image)) / (np.max(image) - np.min(image))

            if i < mask.shape[2]:
                mask_slice = mask[:, :, i]

                # Save only slices with embolism
                if np.max(mask_slice) > 0:
                    img_name = f"{patient}_{i}.png"

                    plt.imsave(f"{output_img}/{img_name}", image, cmap='gray')
                    plt.imsave(f"{output_mask}/{img_name}", mask_slice, cmap='gray')
```

```python
import os
import cv2
import numpy as np

IMG_SIZE = 256

def load_data(image_dir, mask_dir):
    images = []
    masks = []

    file_names = sorted(os.listdir(image_dir))

    for file in file_names:
        print(file)

        img = cv2.imread(os.path.join(image_dir, file), cv2.IMREAD_GRAYSCALE)
        mask = cv2.imread(os.path.join(mask_dir, file), cv2.IMREAD_GRAYSCALE)

        img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
        mask = cv2.resize(mask, (IMG_SIZE, IMG_SIZE))

        img = img / 255.0
        mask = mask / 255.0

        images.append(img)
        masks.append(mask)

    images = np.array(images).reshape(-1, IMG_SIZE, IMG_SIZE, 1)
    masks = np.array(masks).reshape(-1, IMG_SIZE, IMG_SIZE, 1)

    return images, masks
```

```python
import os
import shutil
from sklearn.model_selection import train_test_split

image_dir = "/content/pe_dataset/images"
mask_dir = "/content/pe_dataset/masks"

images = sorted(os.listdir(image_dir))

# Step 1: Train + Temp
train, temp = train_test_split(images, test_size=0.3, random_state=42)

# Step 2: Validation + Test
val, test = train_test_split(temp, test_size=0.5, random_state=42)

# Create folders
base = "/content/dataset_split"

folders = [
    "train/images", "train/masks",
    "val/images", "val/masks",
    "test/images", "test/masks"
]

for f in folders:
    os.makedirs(os.path.join(base, f), exist_ok=True)

# Function to copy
def copy_files(file_list, src_img, src_mask, dst_img, dst_mask):
    for img in file_list:
        shutil.copy(os.path.join(src_img, img), os.path.join(dst_img, img))
        shutil.copy(os.path.join(src_mask, img), os.path.join(dst_mask, img))

# Copy files
copy_files(train, image_dir, mask_dir,
           base + "/train/images", base + "/train/masks")

copy_files(val, image_dir, mask_dir,
           base + "/val/images", base + "/val/masks")

copy_files(test, image_dir, mask_dir,
           base + "/test/images", base + "/test/masks")

print("Train:", len(train))
print("Validation:", len(val))
print("Test:", len(test))
```

```python
train_images, train_masks = load_data(
    "/content/dataset_split/train/images",
    "/content/dataset_split/train/masks"
)

test_images, test_masks = load_data(
    "/content/dataset_split/test/images",
    "/content/dataset_split/test/masks"
)

print(train_images.shape, test_images.shape)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model

def unet():
    inputs = layers.Input((256, 256, 1))

    # Encoder
    c1 = layers.Conv2D(16, 3, activation='relu', padding='same')(inputs)
    p1 = layers.MaxPooling2D()(c1)

    c2 = layers.Conv2D(32, 3, activation='relu', padding='same')(p1)
    p2 = layers.MaxPooling2D()(c2)

    # Bottleneck
    b = layers.Conv2D(64, 3, activation='relu', padding='same')(p2)

    # Decoder
    u1 = layers.UpSampling2D()(b)
    c3 = layers.Conv2D(32, 3, activation='relu', padding='same')(u1)

    u2 = layers.UpSampling2D()(c3)
    c4 = layers.Conv2D(16, 3, activation='relu', padding='same')(u2)

    outputs = layers.Conv2D(1, 1, activation='sigmoid')(c4)

    model = Model(inputs, outputs)
    model.compile(
        optimizer='adam',
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model

model = unet()
model.summary()
```

```python
history = model.fit(
    train_images,
    train_masks,
    validation_data=(test_images, test_masks),
    epochs=10,
    batch_size=8
)
```

```python
loss, acc = model.evaluate(test_images, test_masks)
print("Test Accuracy:", acc)
```
